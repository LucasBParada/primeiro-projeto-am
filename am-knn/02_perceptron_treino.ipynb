```python
import numpy as np

# Dados históricos
X = np.array([
    [1.0,1.0],
    [2.0,1.0],
    [5.0,6.0],
    [6.0,5.0]
])

y = np.array([0,0,1,1])

# 1. Parâmetros iniciais
w = np.zeros(2)
b = 0.0
n = 0.1
epocas = 10

# 2. Treinamento Iterativo
for epoca in range(epocas):
    erros = 0
    for x_i, y_real in zip(X,y):
        # Predição
        y_pred = 1 if (np.dot(x_i, w) + b) >= 0 else 0
        erro = y_real - y_pred

        # Correção caso erre
        if erro !=0:
            w += n * erro * x_i
            b += n * erro
            erro += 1

    if erros == 0:
        print(f"Modelo Convergiu com sucesso na época {epoca + 1}")
        break

print(f"Pesos calibrados: wl = {w[0]:.2f}, w2 = {w[1]:.2f}, b = {b:.2f}")

# 3. Predição para um novo ponto de teste
novo_ponto = np.array([5.5,5.5])
predicao = 1 if (np.dot(novo_ponto, w) + b) >= 0 else 0
rotulo = "Classe 1 (Premium)" if predicao == 1 else "Classe 0 (Básico)"
print(f"Predição para {novo_ponto}: {rotulo}")
```
